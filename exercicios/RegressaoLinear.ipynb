```python
from numpy import *

class LinearRegression:
    def __init__(self, x, y):
      # Inicializa a classe com os dados de entrada (x) e saída (y)
      self.x = x
      self.y = y
      # Calcula os valores necessários para a regressão linear
      self.__correlation_coefficient = self.__corretion()  # Coeficiente de correlação
      self.__inclination = self.__inclination()  # Coeficiente angular (inclinação da reta)
      self.__interception = self.__interception()  # Interceptação (onde a reta cruza o eixo y)

    def __corretion(self):
      # Calcula o coeficiente de correlação de Pearson
      # Este valor mede o grau de relação linear entre as variáveis x e y
      # Varia de -1 a 1, onde:
      # 1: correlação positiva perfeita
      # 0: sem correlação
      # -1: correlação negativa perfeita
      covariation = cov(self.x, self.y, bias=True)[0][1]  # Covariância entre x e y
      variant_x = var(self.x)  # Variância de x
      variant_y = var(self.y)  # Variância de y
      return covariation / sqrt(variant_x * variant_y)

    def __inclination(self):
      # Calcula o coeficiente angular (inclinação) da reta de regressão
      # Usando a fórmula: m = r * (desvio padrão de y / desvio padrão de x)
      # onde r é o coeficiente de correlação
      stdx = std(self.x)  # Desvio padrão de x
      stdy = std(self.y)  # Desvio padrão de y
      return self.__correlation_coefficient * (stdy / stdx)

    def __interception(self):
      # Calcula o valor de interceptação (b) da equação y = mx + b
      # Usando a fórmula: b = média de y - m * média de x
      mediax = mean(self.x)  # Média dos valores de x
      mediay = mean(self.y)  # Média dos valores de y
      return mediay - mediax * self.__inclination

    def prediction(self, valor):
      # Realiza a predição para um novo valor usando a equação da reta y = mx + b
      # onde m é a inclinação e b é a interceptação
      return self.__interception + (self.__inclination * valor)
```

```python
x = array([1,2,3,4,5])
y = array([2,4,6,8,10])

lr = LinearRegression(x,y)
prediction = lr.prediction(6)
print(prediction)
```
